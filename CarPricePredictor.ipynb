```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
data=pd.read_csv("CarPrice.csv")
x=data[["Mileage"]]
y=data["Price"]
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=4)
model=LinearRegression()
model.fit(xtrain,ytrain)
predicty=model.predict(xtest).round(2)
df=pd.DataFrame({'Mileage':xtest["Mileage"].values,'Actual':ytest.values,'Predicted':predicty})
print("Slope:", model.coef_[0])
print("Mean Absolute Error: ",mean_absolute_error(ytest,predicty))
print("Mean Squared Error: ",mean_squared_error(ytest,predicty))
print("R^2 Score: ",r2_score(ytest,predicty))
plt.scatter(x,y,color='r')
x_sorted=xtrain.sort_values(by="Mileage")
plt.plot(x_sorted,model.predict(x_sorted),color='yellow')
plt.xlabel("Mileage")
plt.ylabel("Price")
plt.title("Simple Linear Regression")
plt.show()
print(df.head())
new_data=pd.DataFrame([[70000]],columns=["Mileage"])
pr=model.predict(new_data)

print("Price of Car with 70000 mileage",pr[0])
mileage=float(input("Mileage: "))
new_data=pd.DataFrame([[mileage]],columns=["Mileage"])
pr=model.predict(new_data)

print("Price of Car:",pr[0].round(2))
```
