```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import matplotlib.pyplot as plt
df=pd.read_csv("study.csv")
x=df[["HOURS"]]
y=df["MARKS"]
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.40,random_state=2)
model=LinearRegression()
model.fit(xtrain,ytrain)
predicty=model.predict(xtest)
ytrainpre=model.predict(xtrain)
plt.scatter(x,y,color='yellow')
plt.plot(xtrain,ytrainpre,color='red')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
prediction=pd.DataFrame({'Actual':ytest,'Predicted':predicty})
print(f"Slope: {model.coef_}")
print(f"Intercept: {model.intercept_}")
print(prediction.head())
mse=mean_squared_error(ytest,predicty)
print("Mean Squared Error: ",mse)
r2=r2_score(ytest,predicty)
print("R^2 Score: ",r2)
```
